```python
#두 정수를 큰 수,작은 수 순으로 출력하기
import random
a=random.randint(1,100)
b=random.randint(1,100)
print('교환 전:',a,b)

if a<b:
    a,b=b,a
print('큰 수:',a,'작은 수:',b)
```

```python
#주사위 굴리기
import random
dice1 = random.randint(1, 6)
dice2 = random.randint(1, 6)

print(f'첫 번째 주사위: {dice1}')
print(f'두 번째 주사위: {dice2}')

if dice1 > dice2:
    print('첫 번째 주사위가 큽니다.')
elif dice1 < dice2:
    print('두 번째 주사위가 큽니다.')
else:
    print('두 주사위가 같습니다.')
```

```python
#가위바위보 게임
import random
com=random.randint(1,3)
user=int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == com:
    print('비겼습니다.')
elif user == 1 and com == 3:
    print('이겼습니다!')        
elif user == 2 and com == 1:
    print('이겼습니다!')       
elif user == 3 and com == 2:
    print('이겼습니다!')      
else:
    print('졌습니다...')
```

```python
##랜덤한 숫자 맞추기
import random
com=random.randint(1,100)
user=int(input('생각한 숫자를 입력하시오: '))
print(com)
if com==user:
    print('정답입니다')
elif com<user:
    print('입력한 숫자가 큽니다.')
else:
    print('입력한 숫자가 작습니다.')
```

```python
#함수1개로 사칙연산하기
def asmd(x,y,op):
    if op=='+':
        print(f'{x} + {y} = {x+y}')
    elif op=='-':
        print(f'{x} - {y} = {x-y}')
    elif op=='*':
        print(f'{x} * {y} = {x*y}')
    elif op=='/':
        print(f'{x} / {y} = {x/y:.2f}')
    else:
        print('잘못된 연산자입니다.')
```

```python
x=int(input('첫 번째 수: '))
y=int(input('두 번째 수: '))
op=input('연산자를 입력하시오: ')
asmd(x,y,op)
```

```python
## 반복구조 for문:난수 맞추기
import random
com=random.randint(1,100)

for i in range(5):
    user=int(input('생각한 숫자를 입력하시오: '))
    if com==user:
        print('정답입니다')
        break
    elif com<user:
        print('입력한 숫자가 큽니다.')
    else:
        print('입력한 숫자가 작습니다.')
print(com)
```

```python
## 반복구조 while문:난수 맞추기
import random
com=random.randint(1,100)
count=0

while True:
    user=int(input('생각한 숫자를 입력하시오: '))
    count+=1
    
    if com==user:
        print('정답입니다')
        print(f'정답:{com}')
        print(f'시도 횟수:{count}')
        break
    elif com<user:
        print('입력한 숫자가 큽니다.')
    else:
        print('입력한 숫자가 작습니다.')
```
