```python
#1부터 100까지의 합을 출력하시오
sum=0
for i in range(1,101):
    sum+=i
print(sum)
```

```python
#4-1-1.1부터 100까지의 짝수의 합 구하기
total=0
for n in range(2,101,2):
    total+=n
print(total)
```

```python
#4-1-2.정수 하나를 입력받아 그 수의 구구단 출력하기
n=int(input('단의 정수를 입력하세요: '))
for i in range(1,10):
    print(f'{n}X{i}={n*i}')
```

```python
#2단부터 9단까지 구구단을 7X1=7형식으로 출력하세요
for i in range(2,10):
    for k in range(1,10):
        print(f'{i}x{k}={i*k}')
    print('--------')
```

```python
#while문으로 1부터 100까지의 합을 출력하시오
sum=0
i=1
while i<=100:
    sum+=i
    i+=1
print(sum)
```

```python
#break 반복문 빠져나오기
sum=0
i=1
while True:
    sum+=i
    i+=1
    if i>100:
        break
print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을입력하면 반복을 끝내고 합계 출력
sum=0
while True:
    n=int(input('양의 정수를 입력하시오: '))
    if n==0:
        break
    sum+=n
print('합계:',sum)
```

```python
#5-1.1부터 20까지의 수 중 짝수만 담은 리스트 출력
evens=[]
for i in range(1,21):
    if i%2==0:
        evens.append(i)
print(evens)
evens.pop(7) #pop함수 : 인덱스로 삭제
print(evens)
del evens[5]
print(evens) #del키워드 
```
