```python
#점에 주석추가하기 annotate함수
import matplotlib.pyplot as plt
name = ['Jessica', 'Liam', 'Sophia']
x = [75, 92, 83]
y = [79, 89, 90]

plt.figure(figsize=(5, 3))
plt.scatter(x, y)
for i in range(len(name)):
    plt.annotate(name[i], xy=(x[i], y[i]), xytext=(5, -10), textcoords='offset points')
plt.xlim(73, 95)
plt.ylim(75, 93)
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
#히스토그램
import matplotlib.pyplot as plt
temperature = [6, 2.2, 4.6, 7.7, 8, 9.2, 11.7, 12.5, 10.3, 13.5,
15.2, 5.9, 2.9, 9.2, 8.6, 4.6, 7.2, 7.7, 8.3, 9.5,
13, 17, 17.4, 12.5, 12.5, 9.4, 7.3, 9.2, 11.8, 13.3, 15]
print('일 수:', len(temperature))

plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
count, edges, _ = plt.hist(temperature) 
plt.title('default (10 bins)')
print('구간별 도수:', [int(c) for c in count])

plt.subplot(1, 2, 2)
plt.hist(temperature, bins=15, alpha=0.5) 
plt.xlabel('temperature')
plt.ylabel('frequency')
plt.title('bins=15, alpha=0.5')

plt.tight_layout()
plt.show()
```

```python
##pandas 데이터 프레임 생성
# 1 딕셔너리로 만들기: 키 = 열 이름, 값 = 열의 각 행 데이터(리스트)
import pandas as pd 
df = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
'test': [45, 30, 40, 37, 48],
'assign1': [20, 17, 22, 18, 24],
'assign2': [19, 14, 18, 15, 25]})
print(df)
print()

# 2 리스트 + columns 매개변수로 만들기
df2 = pd.DataFrame([['Jessica', 45, 20, 19], ['Liam', 30, 17, 14], ['Sophia', 40, 22,
18],['Ryan', 37, 18, 15], ['Alex', 48, 24, 25]],
columns=['name', 'test', 'assign1', 'assign2'])

# 3 빈 데이터프레임을 만들고 열을 하나씩 추가하기
df3 = pd.DataFrame()
df3['name'] = ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex']
df3['test'] = [45, 30, 40, 37, 48]
df3['assign1'] = [20, 17, 22, 18, 24]
df3['assign2'] = [19, 14, 18, 15, 25]

print('세 방법의 결과가 같은가?', df.equals(df2), df.equals(df3))
```

```python
#데이터 전처리 : 열이름 변경&정렬
import pandas as pd
df = pd.DataFrame({'name': ['Jessica', 'Liam', 'Sophia', 'Ryan', 'Alex'],
'test': [45, 30, 40, 37, 48],
'assign1': [20, 17, 22, 18, 24],
'assign2': [19, 14, 18, 15, 25]})
df['sum'] = df['test'] + df['assign1'] + df['assign2']

df.rename(columns={'test': 'exam'}, inplace=True) #df 자체가 바뀐다
print(df)

sort_df = df.sort_values('sum', ascending=False)
print(sort_df)
print('원래 df 의 순서는 그대로:', list(df['name']))
```
